## 1️⃣ What is Statistics?

Statistics helps us understand a collection of data.

For example, suppose a company has sales data:

- Revenue
- Quantity
- Region
- Profit

Instead of looking at every row, statistics helps answer:

- What is the **average**?
- What is the **middle value**?
- How much does the data **vary**?
- Are there any **unusual values**?
- What can we learn for a business decision?

### The most important idea

**A number alone is not an insight.**

Example:

> Average Revenue = ₹40,000

The next question should be:

> **So what? What does this mean for the business?**

# 📌 Part A – Measures of Central Tendency

There are three common measures:

### Mean
The normal **average**.

### Median
The **middle** value after sorting.

### Mode
The value that appears most often.

For business data, **mean and median are especially useful to compare**.


## 6️⃣ Mean – The Average

Formula:

**Mean = Sum of all values ÷ Number of values**

In Python:
`mean()`


```python
mean_revenue = df["Revenue"].mean()

print("Average Revenue =", mean_revenue)
```

## 7️⃣ Median – The Middle Value

Imagine sorting all Revenue values from smallest to largest.

The median is the value in the middle.

Why is it useful?

Because a very large or very small value can affect the mean.

So we often compare:

**Mean vs Median**


```python
median_revenue = df["Revenue"].median()

print("Median Revenue =", median_revenue)
```

## 🤔 Think About It

If:

- Mean = ₹45,000
- Median = ₹30,000

What might have happened?

A possible explanation is that **some unusually large orders are pulling the mean upward**.

This is why we should not blindly trust the average.


## 8️⃣ Minimum and Maximum

These tell us the smallest and largest Revenue values.


```python
minimum = df["Revenue"].min()
maximum = df["Revenue"].max()

print("Minimum Revenue =", minimum)
print("Maximum Revenue =", maximum)
```

# 📌 Part B – Variability

Two datasets can have the **same average** but behave very differently.

Example:

### Team A
98, 100, 102, 101, 99

### Team B
40, 60, 100, 140, 160

Both have an average around 100.

But Team B is much more spread out.

That is why we need measures of **variation**.


## 9️⃣ Standard Deviation

Standard deviation tells us how much values tend to vary around the average.

Simple interpretation:

- **Small standard deviation** → values are relatively consistent
- **Large standard deviation** → values are more spread out

It does not tell us whether a business result is "good" or "bad" by itself.


```python
std_revenue = df["Revenue"].std()

print("Standard Deviation =", std_revenue)
```

## 🔎 Compare Mean, Median and Standard Deviation

```python
print("Mean   :", df["Revenue"].mean())
print("Median :", df["Revenue"].median())
print("Std Dev:", df["Revenue"].std())
```

# 📌 Part D – Grouped Statistics

Now let's ask a business question:

> **How are the regions performing?**

Instead of calculating one number for the entire dataset, we can calculate statistics **for each Region**.


```python
region_summary = df.groupby("Region").agg(
    Total_Revenue=("Revenue", "sum"),
    Average_Revenue=("Revenue", "mean"),
    Total_Profit=("Profit", "sum"),
    Orders=("OrderID", "count")
)

region_summary
```

## 🧠 Interpret the Table

Ask these questions:

1. Which region has the highest total Revenue?
2. Which region has the highest average Revenue?
3. Which region has the highest total Profit?
4. Does the highest Revenue region also have the highest Profit?
5. Does one region have very few orders?

### Important

**Highest Revenue does not automatically mean best business performance.**

We need context.


# 📌 Part E – Outliers

An **outlier** is a value that is unusually far from the rest of the observations.

Look at the Revenue column.

One large order may be:

- a genuine bulk order
- a special customer
- a promotion
- a data-entry error

### Key rule

> **An outlier is a question, not automatically an error.**


```python
# Find the largest Revenue values
df[["OrderID", "Region", "Revenue"]].sort_values(
    by="Revenue", ascending=False
).head()
```

## 10️⃣ IQR Method – Simple Version

A common method for identifying possible outliers uses the **Interquartile Range (IQR)**.

- Q1 = 25th percentile
- Q3 = 75th percentile
- IQR = Q3 − Q1

Possible upper outlier:

**Value > Q3 + 1.5 × IQR**


```python
q1 = df["Revenue"].quantile(0.25)
q3 = df["Revenue"].quantile(0.75)

iqr = q3 - q1

lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
```

```python
outliers = df[
    (df["Revenue"] < lower_limit) |
    (df["Revenue"] > upper_limit)
]

outliers
```

# 📌 Part F – Correlation

Suppose we want to know:

> Does Quantity have a relationship with Revenue?

We can calculate correlation.

Correlation is between **-1 and +1**.

- Close to +1 → strong positive linear relationship
- Close to -1 → strong negative linear relationship
- Close to 0 → weak/no linear relationship

### Important

**Correlation does NOT prove causation.**


```python
df[["Quantity", "Revenue", "Profit"]].corr()
```
